# Understanding a URL (for phishing detection)

A link (URL) is an address with several parts, and each part means something.

Example:

```
https://www.google.com/search?q=python
```

## Part by part

| Part | In the example | What it means |
|---|---|---|
| Scheme | `https://` | How the browser connects. `https` is the secure version, `http` is not. |
| Subdomain | `www.` | A section in front of the main name. A site can have many (`mail.`, `login.`, `shop.`). |
| Domain | `google` | The actual name of the website. This is the most important part. |
| TLD | `.com` | The ending: `.com`, `.in`, `.org`, `.xyz`, `.top`. |
| Path | `/search` | Which page or folder on the site. |
| Query | `?q=python` | Extra information sent to the page. |

## The one rule that matters

**The real owner of a link is the domain plus the TLD, just before the first single `/`.**

Read the address from right to left, starting at the TLD:

```
https://login.paypal.com.secure-verify.xyz/account
                              ^^^^^^^^^^^^^^^^^
                       real site: secure-verify.xyz
```

Everything to the left of `secure-verify.xyz` (`login.paypal.com.`) is just a subdomain. The attacker can name it anything, so it only looks like PayPal.

## Safe versus phishing

Safe:

```
https://www.paypal.com/signin
```

The real site is `paypal.com`, and `www` is only a subdomain.

Phishing:

```
http://paypal.com.account-verify.top/login
```

The real site is `account-verify.top`. `paypal.com` is a decoy placed in front of it.

## Why this matters for the model

A model can't "look" at a link the way a person does. It needs numbers, so we split the link into parts and count things: how long it is, how many dots, which TLD, and so on. This is called feature extraction.

## Quick check

What is the real domain in this link, and why?

```
https://secure.bank-login.com.verify-user.xyz/home
```

````python
from urllib.parse import urlparse
import tldextract

url = "http://paypal.com.account-verify.top/login@update?id=123"

parsed = urlparse(url)
ext = tldextract.extract(url)

print(parsed.scheme)   # http
print(parsed.path)     # /login@update
print(parsed.query)    # id=123
print(ext.subdomain)   # paypal.com
print(ext.domain)      # account-verify
print(ext.suffix)      # top
````
